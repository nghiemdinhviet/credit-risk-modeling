```python
from pathlib import Path
import pandas as pd
import numpy as np

CURRENT_DIR = Path.cwd()

if CURRENT_DIR.name == "notebooks":
    PROJECT_ROOT = CURRENT_DIR.parent
else:
    PROJECT_ROOT = CURRENT_DIR

DATA_PATH = PROJECT_ROOT / "data" / "raw" / "credit_default_raw.xls"

print("Project root:", PROJECT_ROOT)
print("Dataset path:", DATA_PATH)
print("File exists:", DATA_PATH.exists())

df = pd.read_excel(DATA_PATH, header=1)

print("\nDataset shape:", df.shape)

df.head()
```

```python
# Kiểm tra cấu trúc tổng quan của dataset

print("Số dòng:", df.shape[0])
print("Số cột:", df.shape[1])

print("\nDanh sách cột:")
for i, col in enumerate(df.columns, start=1):
    print(f"{i:02d}. {col}")

print("\nKiểu dữ liệu:")
df.info()
```

```python
target_col = "default payment next month"

target_summary = pd.DataFrame({
    "Count": df[target_col].value_counts(),
    "Percent": (df[target_col].value_counts(normalize=True) * 100).round(2)
})

target_summary
```

```python
duplicate_count = df.duplicated().sum()

print("Số dòng bị duplicate:", duplicate_count)
print("Tỷ lệ duplicate:", round(duplicate_count / len(df) * 100, 2), "%")
```

```python
print("Số ID duy nhất:", df["ID"].nunique())
print("Tổng số dòng:", len(df))
print("Số ID bị trùng:", df["ID"].duplicated().sum())
```

```python
categorical_cols = ["SEX", "EDUCATION", "MARRIAGE"]

for col in categorical_cols:
    print(f"\n===== {col} =====")
    print(df[col].value_counts().sort_index())
```

```python
payment_status_cols = [
    "PAY_0",
    "PAY_2",
    "PAY_3",
    "PAY_4",
    "PAY_5",
    "PAY_6"
]

for col in payment_status_cols:
    print(f"\n===== {col} =====")
    print(df[col].value_counts().sort_index())
```

```python
df[payment_status_cols].agg(["min", "max"])
```

```python
for col in payment_status_cols:
    summary = (
        df.groupby(col)[target_col]
        .agg(
            Customers="count",
            Defaults="sum",
            Default_Rate="mean"
        )
    )

    summary["Default_Rate"] = (
        summary["Default_Rate"] * 100
    ).round(2)

    print(f"\n===== {col} =====")
    print(summary)
```

```python
numeric_check_cols = [
    "LIMIT_BAL",
    "AGE",
    "BILL_AMT1",
    "BILL_AMT2",
    "BILL_AMT3",
    "BILL_AMT4",
    "BILL_AMT5",
    "BILL_AMT6",
    "PAY_AMT1",
    "PAY_AMT2",
    "PAY_AMT3",
    "PAY_AMT4",
    "PAY_AMT5",
    "PAY_AMT6"
]

df[numeric_check_cols].describe().T[
    ["min", "mean", "50%", "max"]
].round(2)
```

## Data Understanding Summary

- Dataset contains 30,000 customers and 25 variables.
- No missing values were detected.
- No full-row duplicates were detected.
- Target variable:
  - Non-default: 23,364 customers (77.88%)
  - Default: 6,636 customers (22.12%)
- The target is moderately imbalanced, so accuracy alone will not be sufficient for model evaluation.
- SEX contains valid categories only.
- EDUCATION contains undocumented categories 0, 5, and 6.
- MARRIAGE contains undocumented category 0.
- Payment status variables PAY_0 to PAY_6 range from -2 to 8.
- Payment delay history shows a strong relationship with default risk.
- Negative BILL_AMT values will not be removed without further investigation.
- Large payment and bill values will be examined during EDA before any outlier treatment.
